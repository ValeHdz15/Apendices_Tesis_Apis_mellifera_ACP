```python
#Fast filtering according to literature
# Step 1: Remove those peptides that do not contain lysine or arginine
# Step 2: Remove those peptides that contains more arginines than lysines
with open("toxi sin fasta.txt") as input_file:
  list_peptides = input_file.readlines()

output_file = open("Results_steps_1_2.txt", "w")

num_peptides = 0

for line in list_peptides:
  line = line.strip("\n")
  numK = line.upper().count("K")
  numR = line.upper().count("R")
  if(numK>0 or numR>0):
    if(numR<numK):
      num_peptides += 1
      str_line = line + "\n"
      output_file.write(str_line)

output_file.close()
print("Number of selected peptides: ", num_peptides)

```

```python
#FILTRANDO POR PERCENTILES 10 - 90
# Step 3. Select those peptides with their physicochemical properties in the ACP ranges
# ACP ranges were analyzed from those peptides from the DBAASP database
!pip install peptides
import peptides
import os

with open("SecuenciasAFiltrarApis.txt") as input_file:
  data = input_file.readlines()

output_file = open("Results_step_3.txt", "w")

num_peptides = 0

for line in data:
    line = line.strip("\n")
    peptide = peptides.Peptide(line)

    charge = peptide.charge(pH=7.4)
    if(charge>=-0.005 and charge<=5.993):
      iso_point = peptide.isoelectric_point()
      if(iso_point>=6.100 and iso_point<12.977):
        alip_index = peptide.aliphatic_index()
        if(alip_index>=0.000 and alip_index<=186.222):
          inst_index = peptide.instability_index()
          if(inst_index>=-16.892 and inst_index<=148.255):
            hydro_moment= peptide.hydrophobic_moment()
            if(hydro_moment>=0.178 and hydro_moment<=0.885):
              hydrophob = peptide.hydrophobicity()
              if(hydrophob>=-1.865 and hydrophob<=1.337):
                num_peptides += 1
                str_line = line + "\n"
                output_file.write(str_line)
                print(line)

print("Total number of used peptides: ", num_peptides)
output_file.close()
```

```python
#Analysis of the candidate peptides
!pip install peptides
import peptides
import os

with open("Results_step_3.txt") as input_file:
  data = input_file.readlines()

output_file = open("Res_Analysis_FQ.txt", "w")
str_line = "Sequence \t aliphatic index \t charge \t hyd moment \t hydrophob \t instability \t ip \n"
output_file.write(str_line)

num_peptides = 0

for line in data:
    line = line.strip("\n")
    peptide = peptides.Peptide(line)

    charge = peptide.charge(pH=7.4)
    iso_point = peptide.isoelectric_point()
    alip_index = peptide.aliphatic_index()
    inst_index = peptide.instability_index()
    hydro_moment= peptide.hydrophobic_moment()
    hydrophob = peptide.hydrophobicity()

    "Sequence \t aliphatic index \t charge \t hyd moment \t hydrophob \t instability \t ip \n"
    str_line = line + "\t" + str(alip_index) + "\t" + str(charge) + "\t" + str(hydro_moment)
    str_line += "\t" + str(hydrophob) + "\t" + str(inst_index) + "\t" + str(iso_point) + "\n"
    output_file.write(str_line)
    print(str_line)

output_file.close()
```
