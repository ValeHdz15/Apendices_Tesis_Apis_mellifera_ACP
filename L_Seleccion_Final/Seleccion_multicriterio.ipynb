```python
import re

#SISTEMA SELECCIÓN MULTICRITERIO

with open("83.csv") as input_file:
  data = input_file.readlines()

numSelected = 0

for line in data:
  values = line.split(";")
  numReq = 0

  #Anticp2
  if (float(values[2].replace(',', '.'))>=0.95):
    numReq += 2

  if(float(values[4].replace(',', '.'))>=0.95):
    numReq += 2

  #ToxinPred
  if(float(values[3].replace(',', '.'))<=0.35):
    numReq += 1


  #2D Structure
  cadena = values[5]
  subcadena = cadena[1:-1]
  numE = subcadena.count("E")
  numH = subcadena.count("H")
  numC = subcadena.count("C")
  numG = subcadena.count("G")
  if(numE>=numC or numH>=numC or numG>=numC):
    numReq += 2

  #Peptide Cutter
  pepCutter = values[6].strip()
  if(int(pepCutter)<2):
    numReq += 1

  # Ramachandran
  totalRes = float(values[7].strip().replace(',', '.'))
  favorecidos = float(values[8].strip().replace(',', '.'))

  if(totalRes>=4 and favorecidos>=90):
    numReq += 2
  elif(favorecidos >=70):
    numReq += 1


  #Print selected
  if(numReq>=8):
    print(values)
    numSelected += 1
```
