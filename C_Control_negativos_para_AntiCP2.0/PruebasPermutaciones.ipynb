```python
#Prueba_Barajar_Secuencias
import random

def shuffle_seq(seq):
    seq = list(seq)
    random.shuffle(seq)
    return "".join(seq)

with open("Archivo:referencia.txt") as f, \
     open("shuf1.txt","w") as f1, \
     open("shuf2.txt","w") as f2, \
     open("shuf3.txt","w") as f3, \
     open("shuf4.txt","w") as f4, \
     open("shuf5.txt","w") as f5, \
     open("shuf6.txt","w") as f6, \
     open("shuf7.txt","w") as f7, \
     open("shuf8.txt","w") as f8, \
     open("shuf9.txt","w") as f9, \
     open("shuf10.txt","w") as f10:

    outs = [f1,f2,f3,f4,f5,f6,f7,f8,f9,f10]

    for pep in f:
        pep = pep.strip()
        if not pep:
            continue

        for out in outs:
            out.write(shuffle_seq(pep) + "\n")

print("Listo")
```

```python
# Convertir los 10 archivos TXT a formato FASTA
# NO se modifican las secuencias

for i in range(1, 11):

    archivo_txt = f"shuf{i}.txt"
    archivo_fasta = f"shuf{i}.fasta"

    with open(archivo_txt, "r") as entrada, \
         open(archivo_fasta, "w") as salida:

        contador = 0

        for linea in entrada:
            secuencia = linea.strip()

            # Ignorar líneas vacías
            if not secuencia:
                continue

            contador += 1

            # Identificador FASTA
            salida.write(f">Sec_{contador}\n")

            # Secuencia sin modificar
            salida.write(secuencia + "\n")

    print(f"{archivo_txt} → {archivo_fasta}: {contador} secuencias")

print("Conversión a FASTA terminada.")
```
